```python
# =========================================================
# HYBRID IDS: PSO + NEGATIVE SELECTION AIS + MLP + RF + XGBOOST + ENSEMBLE
# Dataset: NSL_KDD
# FULL VERSION WITH 10-FOLD CROSS-VALIDATION
# =========================================================

# =========================================================
# Step 1: Import Libraries
# =========================================================
import os
import random
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore")

from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    balanced_accuracy_score,
    roc_curve,
    auc,
    precision_recall_curve
)
from sklearn.base import clone

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping

# Install xgboost in Colab if needed
try:
    from xgboost import XGBClassifier
except:
    !pip install -q xgboost
    from xgboost import XGBClassifier

# =========================================================
# Reproducibility
# =========================================================
SEED = 42
np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)

sns.set(style="whitegrid")
print("✅ Libraries imported successfully!")

# =========================================================
# Step 2: Mount Google Drive
# =========================================================
from google.colab import drive
drive.mount('/content/drive')

# =========================================================
# Step 3: Load NSL_KDD Dataset
# =========================================================
dataset_path = '/content/drive/MyDrive/Computational_Intelligence/NSL_KDD'

files = os.listdir(dataset_path)
print("Available files:", files)

train_file = None
test_file = None

for f in files:
    if 'train' in f.lower() and f.endswith('.csv'):
        train_file = f
    elif 'test' in f.lower() and f.endswith('.csv'):
        test_file = f

if train_file is None or test_file is None:
    raise FileNotFoundError("Train/Test CSV files not found in the NSL_KDD folder.")

train_df = pd.read_csv(os.path.join(dataset_path, train_file))
test_df = pd.read_csv(os.path.join(dataset_path, test_file))

print("Train shape:", train_df.shape)
print("Test shape :", test_df.shape)

# =========================================================
# Step 4: Assign Column Names If Needed
# Uncomment only if your CSV files have no header
# =========================================================
"""
column_names = [
    'duration', 'protocol_type', 'service', 'flag', 'src_bytes', 'dst_bytes',
    'land', 'wrong_fragment', 'urgent', 'hot', 'num_failed_logins',
    'logged_in', 'num_compromised', 'root_shell', 'su_attempted', 'num_root',
    'num_file_creations', 'num_shells', 'num_access_files', 'num_outbound_cmds',
    'is_host_login', 'is_guest_login', 'count', 'srv_count', 'serror_rate',
    'srv_serror_rate', 'rerror_rate', 'srv_rerror_rate', 'same_srv_rate',
    'diff_srv_rate', 'srv_diff_host_rate', 'dst_host_count', 'dst_host_srv_count',
    'dst_host_same_srv_rate', 'dst_host_diff_srv_rate', 'dst_host_same_src_port_rate',
    'dst_host_srv_diff_host_rate', 'dst_host_serror_rate', 'dst_host_srv_serror_rate',
    'dst_host_rerror_rate', 'dst_host_srv_rerror_rate', 'label', 'difficulty'
]
train_df.columns = column_names
test_df.columns = column_names
"""

# =========================================================
# Step 5: Clean Labels and Optional Columns
# =========================================================
if 'difficulty' in train_df.columns:
    train_df = train_df.drop(columns=['difficulty'])
if 'difficulty' in test_df.columns:
    test_df = test_df.drop(columns=['difficulty'])

if 'label' not in train_df.columns:
    train_df.rename(columns={train_df.columns[-1]: 'label'}, inplace=True)
    test_df.rename(columns={test_df.columns[-1]: 'label'}, inplace=True)

train_df['label'] = train_df['label'].astype(str)
test_df['label'] = test_df['label'].astype(str)

# Binary conversion: normal=0, attack=1
train_df['binary_label'] = train_df['label'].apply(lambda x: 0 if x in ['normal', 'normal.'] else 1)
test_df['binary_label'] = test_df['label'].apply(lambda x: 0 if x in ['normal', 'normal.'] else 1)

print("\nTrain binary distribution:")
print(train_df['binary_label'].value_counts())

print("\nTest binary distribution:")
print(test_df['binary_label'].value_counts())

# =========================================================
# Step 6: Encode Categorical Features
# =========================================================
categorical_cols = ['protocol_type', 'service', 'flag']
existing_cat_cols = [col for col in categorical_cols if col in train_df.columns]

combined_df = pd.concat(
    [train_df.drop(columns=['binary_label']), test_df.drop(columns=['binary_label'])],
    axis=0
).reset_index(drop=True)

for col in existing_cat_cols:
    le = LabelEncoder()
    combined_df[col] = le.fit_transform(combined_df[col].astype(str))

train_processed = combined_df.iloc[:len(train_df)].copy()
test_processed = combined_df.iloc[len(train_df):].copy()

# =========================================================
# Step 7: Separate Features and Labels
# =========================================================
X_train_full = train_processed.drop(columns=['label'])
X_test_full = test_processed.drop(columns=['label'])

y_train_full = train_df['binary_label'].values
y_test = test_df['binary_label'].values

X_train_full = X_train_full.select_dtypes(include=[np.number]).copy()
X_test_full = X_test_full.select_dtypes(include=[np.number]).copy()

# =========================================================
# Step 7.1: Remove Infinite and Missing Values
# =========================================================
X_train_full = X_train_full.replace([np.inf, -np.inf], np.nan)
X_test_full = X_test_full.replace([np.inf, -np.inf], np.nan)

train_mask = ~X_train_full.isna().any(axis=1)
test_mask = ~X_test_full.isna().any(axis=1)

X_train_full = X_train_full[train_mask].reset_index(drop=True)
y_train_full = y_train_full[train_mask]

X_test_full = X_test_full[test_mask].reset_index(drop=True)
y_test = y_test[test_mask]

# =========================================================
# Step 8: Standardization (Z-score)
# NOTE:
# We keep your original structure. For strict CV purity, scaling should be
# fitted inside each fold, but here we preserve your pipeline style.
# =========================================================
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_full)
X_test_scaled = scaler.transform(X_test_full)

X_train_full = pd.DataFrame(X_train_scaled, columns=X_train_full.columns)
X_test_full = pd.DataFrame(X_test_scaled, columns=X_test_full.columns)

print("\nPreprocessed training shape:", X_train_full.shape)
print("Preprocessed test shape    :", X_test_full.shape)

# =========================================================
# Step 9: Hold-Out Split ONLY for PSO + Threshold Tuning + Detector Building
# We keep a separate validation split to perform:
# - PSO fitness evaluation
# - MLP early stopping
# - Threshold tuning
# The final model robustness is then checked using 10-fold CV on hybrid features.
# =========================================================
X_train_base, X_val_base, y_train_base, y_val_base = train_test_split(
    X_train_full,
    y_train_full,
    test_size=0.2,
    random_state=SEED,
    stratify=y_train_full
)

X_train_base = X_train_base.reset_index(drop=True)
X_val_base = X_val_base.reset_index(drop=True)

print("\nBase train:", X_train_base.shape)
print("Base val  :", X_val_base.shape)
print("Test      :", X_test_full.shape)

# =========================================================
# Step 10: Balanced Subset for PSO
# =========================================================
def create_balanced_subset(X, y, samples_per_class=4000):
    normal_idx = np.where(y == 0)[0]
    attack_idx = np.where(y == 1)[0]

    n_normal = min(len(normal_idx), samples_per_class)
    n_attack = min(len(attack_idx), samples_per_class)

    chosen_normal = np.random.choice(normal_idx, n_normal, replace=False)
    chosen_attack = np.random.choice(attack_idx, n_attack, replace=False)

    chosen_idx = np.concatenate([chosen_normal, chosen_attack])
    np.random.shuffle(chosen_idx)

    X_sub = X.iloc[chosen_idx].reset_index(drop=True)
    y_sub = y[chosen_idx]
    return X_sub, y_sub

X_pso, y_pso = create_balanced_subset(X_train_base, y_train_base, samples_per_class=4000)

# =========================================================
# Step 11: PSO Feature Selection using RF Fitness
# =========================================================
X_pso_np = X_pso.values
X_val_np = X_val_base.values

def fitness_function(particle, X_train_sub, X_val_full, y_train_sub, y_val, alpha=0.97):
    selected = np.where(particle == 1)[0]

    if len(selected) == 0:
        return 0.0

    Xtr = X_train_sub[:, selected]
    Xv = X_val_full[:, selected]

    clf = RandomForestClassifier(
        n_estimators=120,
        random_state=SEED,
        n_jobs=-1
    )
    clf.fit(Xtr, y_train_sub)
    preds = clf.predict(Xv)

    bal_acc = balanced_accuracy_score(y_val, preds)
    feature_ratio = len(selected) / X_train_sub.shape[1]
    fitness = alpha * bal_acc + (1 - alpha) * (1 - feature_ratio)
    return fitness

num_features = X_pso_np.shape[1]
num_particles = 20
num_iterations = 25
w, c1, c2 = 0.72, 1.49, 1.49

particles = np.random.randint(0, 2, size=(num_particles, num_features))
velocities = np.random.uniform(-1, 1, size=(num_particles, num_features))

for i in range(num_particles):
    if particles[i].sum() == 0:
        particles[i, np.random.randint(0, num_features)] = 1

pbest_positions = particles.copy()
pbest_scores = np.zeros(num_particles)
gbest_position = None
gbest_score = -1
gbest_history = []

for i in range(num_particles):
    score = fitness_function(particles[i], X_pso_np, X_val_np, y_pso, y_val_base)
    pbest_scores[i] = score
    if score > gbest_score:
        gbest_score = score
        gbest_position = particles[i].copy()

print("\nStarting PSO...")
for iteration in range(num_iterations):
    for i in range(num_particles):
        r1 = np.random.rand(num_features)
        r2 = np.random.rand(num_features)

        velocities[i] = (
            w * velocities[i]
            + c1 * r1 * (pbest_positions[i] - particles[i])
            + c2 * r2 * (gbest_position - particles[i])
        )

        sigmoid = 1 / (1 + np.exp(-velocities[i]))
        particles[i] = (np.random.rand(num_features) < sigmoid).astype(int)

        if particles[i].sum() == 0:
            particles[i, np.random.randint(0, num_features)] = 1

        score = fitness_function(particles[i], X_pso_np, X_val_np, y_pso, y_val_base)

        if score > pbest_scores[i]:
            pbest_scores[i] = score
            pbest_positions[i] = particles[i].copy()

        if score > gbest_score:
            gbest_score = score
            gbest_position = particles[i].copy()

    gbest_history.append(gbest_score)
    print(f"Iteration {iteration + 1}/{num_iterations}, Best Fitness: {gbest_score:.4f}")

selected_feature_indices = np.where(gbest_position == 1)[0]
selected_feature_names = X_train_base.columns[selected_feature_indices]

print("\nSelected feature count:", len(selected_feature_indices))
print("Selected features:", list(selected_feature_names))

# =========================================================
# Step 12: Apply Selected Features to Full Training and Test
# =========================================================
X_train_sel_full = X_train_full.values[:, selected_feature_indices]
X_test_sel = X_test_full.values[:, selected_feature_indices]

X_train_sel_base = X_train_base.values[:, selected_feature_indices]
X_val_sel_base = X_val_base.values[:, selected_feature_indices]

# =========================================================
# Step 13: Balanced AIS Training Data
# =========================================================
def create_balanced_numpy_subset(X, y, normal_count=5000, attack_count=5000):
    normal_idx = np.where(y == 0)[0]
    attack_idx = np.where(y == 1)[0]

    n_normal = min(len(normal_idx), normal_count)
    n_attack = min(len(attack_idx), attack_count)

    chosen_normal = np.random.choice(normal_idx, n_normal, replace=False)
    chosen_attack = np.random.choice(attack_idx, n_attack, replace=False)

    return X[chosen_normal], X[chosen_attack]

self_samples, attack_samples = create_balanced_numpy_subset(
    X_train_sel_base, y_train_base, normal_count=5000, attack_count=5000
)

print("\nSelf samples :", self_samples.shape)
print("Attack samples:", attack_samples.shape)

# =========================================================
# Step 14: Negative Selection Detector Generation
# =========================================================
def generate_attack_detectors(self_samples, attack_samples, num_detectors=1000, min_self_distance=0.6):
    detectors = []
    attempts = 0
    max_attempts = num_detectors * 100

    while len(detectors) < num_detectors and attempts < max_attempts:
        base = attack_samples[np.random.randint(0, len(attack_samples))].copy()
        candidate = base + np.random.normal(0, 0.08, size=base.shape)

        d_self = np.min(np.linalg.norm(self_samples - candidate, axis=1))

        if d_self >= min_self_distance:
            detectors.append(candidate)

        attempts += 1

    return np.array(detectors)

detectors = generate_attack_detectors(
    self_samples=self_samples,
    attack_samples=attack_samples,
    num_detectors=1000,
    min_self_distance=0.6
)

print("\nDetector shape:", detectors.shape)

# =========================================================
# Step 15: Create AIS Anomaly Features
# =========================================================
def create_ais_scores(X, detectors):
    feature_list = []

    for sample in X:
        dists = np.linalg.norm(detectors - sample, axis=1)
        sorted_dists = np.sort(dists)

        min_dist = np.min(dists)
        mean_dist = np.mean(dists)
        std_dist = np.std(dists)
        median_dist = np.median(dists)

        closest3_mean = np.mean(sorted_dists[:3]) if len(sorted_dists) >= 3 else np.mean(sorted_dists)
        closest5_mean = np.mean(sorted_dists[:5]) if len(sorted_dists) >= 5 else np.mean(sorted_dists)
        closest10_mean = np.mean(sorted_dists[:10]) if len(sorted_dists) >= 10 else np.mean(sorted_dists)

        activated_08 = np.sum(dists <= 0.8)
        activated_10 = np.sum(dists <= 1.0)
        activated_15 = np.sum(dists <= 1.5)

        affinities = 1 / (1 + dists)
        max_affinity = np.max(affinities)
        mean_affinity = np.mean(affinities)
        std_affinity = np.std(affinities)

        feature_list.append([
            min_dist,
            mean_dist,
            std_dist,
            median_dist,
            closest3_mean,
            closest5_mean,
            closest10_mean,
            activated_08,
            activated_10,
            activated_15,
            max_affinity,
            mean_affinity,
            std_affinity
        ])

    return np.array(feature_list)

# Full training and test AIS features
X_train_ais_full = create_ais_scores(X_train_sel_full, detectors)
X_test_ais = create_ais_scores(X_test_sel, detectors)

# Base split AIS features for threshold tuning / early stopping
X_train_ais_base = create_ais_scores(X_train_sel_base, detectors)
X_val_ais_base = create_ais_scores(X_val_sel_base, detectors)

print("\nAIS feature shape (train full):", X_train_ais_full.shape)

# =========================================================
# Step 16: Merge Selected Features + AIS Features
# =========================================================
X_train_hybrid_full = np.hstack((X_train_sel_full, X_train_ais_full))
X_test_hybrid = np.hstack((X_test_sel, X_test_ais))

X_train_hybrid_base = np.hstack((X_train_sel_base, X_train_ais_base))
X_val_hybrid_base = np.hstack((X_val_sel_base, X_val_ais_base))

print("Hybrid full train shape:", X_train_hybrid_full.shape)
print("Hybrid test shape      :", X_test_hybrid.shape)

# =========================================================
# Step 17: Evaluation Helpers
# =========================================================
results_summary = []

def compute_metrics(y_true, y_pred, y_prob):
    cm = confusion_matrix(y_true, y_pred)
    TN, FP, FN, TP = cm.ravel()

    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)
    bal_acc = balanced_accuracy_score(y_true, y_pred)
    fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
    tnr = TN / (TN + FP) if (TN + FP) > 0 else 0

    fpr_curve, tpr_curve, _ = roc_curve(y_true, y_prob)
    roc_auc = auc(fpr_curve, tpr_curve)

    precision_curve, recall_curve, _ = precision_recall_curve(y_true, y_prob)
    pr_auc = auc(recall_curve, precision_curve)

    return {
        'cm': cm,
        'TN': TN, 'FP': FP, 'FN': FN, 'TP': TP,
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1': f1,
        'balanced_accuracy': bal_acc,
        'fpr': fpr,
        'tnr': tnr,
        'roc_fpr': fpr_curve,
        'roc_tpr': tpr_curve,
        'roc_auc': roc_auc,
        'pr_precision': precision_curve,
        'pr_recall': recall_curve,
        'pr_auc': pr_auc
    }

def print_metrics(name, metrics_dict):
    print(f"\n{'='*70}")
    print(name)
    print(f"{'='*70}")
    print(f"Accuracy          : {metrics_dict['accuracy']:.4f}")
    print(f"Precision         : {metrics_dict['precision']:.4f}")
    print(f"Recall            : {metrics_dict['recall']:.4f}")
    print(f"F1-score          : {metrics_dict['f1']:.4f}")
    print(f"Balanced Accuracy : {metrics_dict['balanced_accuracy']:.4f}")
    print(f"False Positive Rate (FPR): {metrics_dict['fpr']:.4f}")
    print(f"True Negative Rate (TNR) : {metrics_dict['tnr']:.4f}")
    print("\nConfusion Matrix:")
    print(metrics_dict['cm'])

def plot_confusion_matrix(cm, title):
    plt.figure(figsize=(6, 5))
    sns.heatmap(
        cm, annot=True, fmt='d', cmap='Blues',
        xticklabels=['Normal', 'Attack'],
        yticklabels=['Normal', 'Attack']
    )
    plt.title(title, fontsize=13, fontweight='bold')
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")
    plt.tight_layout()
    plt.show()

def add_results_row(name, metrics_dict):
    results_summary.append({
        'Model': name,
        'Accuracy': metrics_dict['accuracy'],
        'Precision': metrics_dict['precision'],
        'Recall': metrics_dict['recall'],
        'F1-score': metrics_dict['f1'],
        'Balanced Accuracy': metrics_dict['balanced_accuracy'],
        'False Positive Rate': metrics_dict['fpr'],
        'True Negative Rate': metrics_dict['tnr'],
        'ROC-AUC': metrics_dict['roc_auc'],
        'PR-AUC': metrics_dict['pr_auc']
    })

# =========================================================
# Step 18: 10-Fold Cross-Validation Helpers
# =========================================================
def create_mlp_model(input_dim):
    model = Sequential([
        Input(shape=(input_dim,)),
        Dense(256, activation='relu'),
        Dropout(0.4),
        Dense(128, activation='relu'),
        Dropout(0.3),
        Dense(64, activation='relu'),
        Dropout(0.2),
        Dense(32, activation='relu'),
        Dense(1, activation='sigmoid')
    ])

    model.compile(
        optimizer='adam',
        loss='binary_crossentropy',
        metrics=['accuracy']
    )
    return model

def run_cv_for_sklearn_model(model, X, y, model_name, n_splits=10):
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=SEED)

    fold_rows = []
    oof_probs = np.zeros(len(y), dtype=float)
    oof_preds = np.zeros(len(y), dtype=int)

    print(f"\n{'='*80}")
    print(f"10-FOLD CV: {model_name}")
    print(f"{'='*80}")

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y), start=1):
        print(f"\nFold {fold}/{n_splits}")

        X_tr, X_va = X[train_idx], X[val_idx]
        y_tr, y_va = y[train_idx], y[val_idx]

        clf = clone(model)
        clf.fit(X_tr, y_tr)

        probs = clf.predict_proba(X_va)[:, 1]
        preds = (probs >= 0.5).astype(int)

        oof_probs[val_idx] = probs
        oof_preds[val_idx] = preds

        fold_metrics = compute_metrics(y_va, preds, probs)
        fold_rows.append({
            'Fold': fold,
            'Accuracy': fold_metrics['accuracy'],
            'Precision': fold_metrics['precision'],
            'Recall': fold_metrics['recall'],
            'F1-score': fold_metrics['f1'],
            'Balanced Accuracy': fold_metrics['balanced_accuracy'],
            'ROC-AUC': fold_metrics['roc_auc'],
            'PR-AUC': fold_metrics['pr_auc']
        })

        print(
            f"Accuracy={fold_metrics['accuracy']:.4f}, "
            f"Precision={fold_metrics['precision']:.4f}, "
            f"Recall={fold_metrics['recall']:.4f}, "
            f"F1={fold_metrics['f1']:.4f}, "
            f"ROC-AUC={fold_metrics['roc_auc']:.4f}"
        )

    cv_df = pd.DataFrame(fold_rows)
    mean_row = cv_df.drop(columns=['Fold']).mean().to_dict()
    std_row = cv_df.drop(columns=['Fold']).std().to_dict()

    print(f"\n{model_name} CV Mean:")
    print(pd.Series(mean_row).round(4))
    print(f"\n{model_name} CV Std:")
    print(pd.Series(std_row).round(4))

    return cv_df, mean_row, std_row, oof_probs, oof_preds

def run_cv_for_mlp(X, y, model_name="MLP", n_splits=10, epochs=80, batch_size=128):
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=SEED)

    fold_rows = []
    oof_probs = np.zeros(len(y), dtype=float)
    oof_preds = np.zeros(len(y), dtype=int)
    histories = []

    print(f"\n{'='*80}")
    print(f"10-FOLD CV: {model_name}")
    print(f"{'='*80}")

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y), start=1):
        print(f"\nFold {fold}/{n_splits}")

        X_tr, X_va = X[train_idx], X[val_idx]
        y_tr, y_va = y[train_idx], y[val_idx]

        tf.keras.backend.clear_session()
        model = create_mlp_model(X.shape[1])

        early_stop = EarlyStopping(
            monitor='val_loss',
            patience=6,
            restore_best_weights=True
        )

        history = model.fit(
            X_tr, y_tr,
            validation_data=(X_va, y_va),
            epochs=epochs,
            batch_size=batch_size,
            callbacks=[early_stop],
            verbose=0
        )
        histories.append(history.history)

        probs = model.predict(X_va, verbose=0).ravel()
        preds = (probs >= 0.5).astype(int)

        oof_probs[val_idx] = probs
        oof_preds[val_idx] = preds

        fold_metrics = compute_metrics(y_va, preds, probs)
        fold_rows.append({
            'Fold': fold,
            'Accuracy': fold_metrics['accuracy'],
            'Precision': fold_metrics['precision'],
            'Recall': fold_metrics['recall'],
            'F1-score': fold_metrics['f1'],
            'Balanced Accuracy': fold_metrics['balanced_accuracy'],
            'ROC-AUC': fold_metrics['roc_auc'],
            'PR-AUC': fold_metrics['pr_auc']
        })

        print(
            f"Accuracy={fold_metrics['accuracy']:.4f}, "
            f"Precision={fold_metrics['precision']:.4f}, "
            f"Recall={fold_metrics['recall']:.4f}, "
            f"F1={fold_metrics['f1']:.4f}, "
            f"ROC-AUC={fold_metrics['roc_auc']:.4f}"
        )

    cv_df = pd.DataFrame(fold_rows)
    mean_row = cv_df.drop(columns=['Fold']).mean().to_dict()
    std_row = cv_df.drop(columns=['Fold']).std().to_dict()

    print(f"\n{model_name} CV Mean:")
    print(pd.Series(mean_row).round(4))
    print(f"\n{model_name} CV Std:")
    print(pd.Series(std_row).round(4))

    return cv_df, mean_row, std_row, oof_probs, oof_preds, histories

# =========================================================
# Step 19: Define Models for CV
# =========================================================
rf_model = RandomForestClassifier(
    n_estimators=400,
    max_depth=None,
    random_state=SEED,
    n_jobs=-1
)

xgb_model = XGBClassifier(
    n_estimators=900,
    max_depth=10,
    learning_rate=0.03,
    subsample=0.90,
    colsample_bytree=0.90,
    gamma=0.1,
    min_child_weight=2,
    objective='binary:logistic',
    eval_metric='logloss',
    random_state=SEED,
    n_jobs=-1
)

# =========================================================
# Step 20: 10-Fold CV on Hybrid Features
# =========================================================
X_cv = X_train_hybrid_full
y_cv = y_train_full

mlp_cv_df, mlp_cv_mean, mlp_cv_std, mlp_oof_probs, mlp_oof_preds, mlp_histories = run_cv_for_mlp(
    X_cv, y_cv, model_name="MLP", n_splits=10, epochs=80, batch_size=128
)

rf_cv_df, rf_cv_mean, rf_cv_std, rf_oof_probs, rf_oof_preds = run_cv_for_sklearn_model(
    rf_model, X_cv, y_cv, model_name="Random Forest", n_splits=10
)

xgb_cv_df, xgb_cv_mean, xgb_cv_std, xgb_oof_probs, xgb_oof_preds = run_cv_for_sklearn_model(
    xgb_model, X_cv, y_cv, model_name="XGBoost", n_splits=10
)

# =========================================================
# Step 21: CV-Based Ensemble Performance on Training Set
# Using out-of-fold predictions to estimate ensemble CV performance
# =========================================================
hard_vote_oof = mlp_oof_preds + rf_oof_preds + xgb_oof_preds
hard_ensemble_oof_preds = (hard_vote_oof >= 2).astype(int)
hard_ensemble_oof_probs = (mlp_oof_probs + rf_oof_probs + xgb_oof_probs) / 3.0

hard_cv_metrics = compute_metrics(y_cv, hard_ensemble_oof_preds, hard_ensemble_oof_probs)
print_metrics("OOF HARD VOTING ENSEMBLE (CV ESTIMATE)", hard_cv_metrics)

soft_oof_probs = (
    0.15 * mlp_oof_probs +
    0.30 * rf_oof_probs +
    0.55 * xgb_oof_probs
)
soft_oof_preds = (soft_oof_probs >= 0.5).astype(int)

soft_cv_metrics = compute_metrics(y_cv, soft_oof_preds, soft_oof_probs)
print_metrics("OOF SOFT VOTING ENSEMBLE (CV ESTIMATE)", soft_cv_metrics)

# =========================================================
# Step 22: Plot Average MLP Training Curve from CV
# =========================================================
max_epochs = max(len(h['loss']) for h in mlp_histories)

avg_train_loss = np.zeros(max_epochs)
avg_val_loss = np.zeros(max_epochs)
count_train = np.zeros(max_epochs)
count_val = np.zeros(max_epochs)

for h in mlp_histories:
    for i, val in enumerate(h['loss']):
        avg_train_loss[i] += val
        count_train[i] += 1
    for i, val in enumerate(h['val_loss']):
        avg_val_loss[i] += val
        count_val[i] += 1

avg_train_loss = avg_train_loss / np.maximum(count_train, 1)
avg_val_loss = avg_val_loss / np.maximum(count_val, 1)

plt.figure(figsize=(8, 5))
plt.plot(avg_train_loss, label='Average Training Loss', linewidth=2)
plt.plot(avg_val_loss, label='Average Validation Loss', linewidth=2)
plt.title('Average Training vs Validation Loss Across 10 Folds (MLP)', fontsize=13, fontweight='bold')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.tight_layout()
plt.show()

# =========================================================
# Step 23: Final Training on Full Hybrid Training Set
# Then evaluate on untouched test set
# =========================================================
candidate_thresholds = np.arange(0.30, 0.71, 0.02)

# ---------- Final MLP ----------
tf.keras.backend.clear_session()
final_mlp = create_mlp_model(X_train_hybrid_full.shape[1])

early_stop = EarlyStopping(
    monitor='val_loss',
    patience=6,
    restore_best_weights=True
)

# keep a small validation from train for threshold tuning / early stopping
X_train_final, X_val_final, y_train_final, y_val_final = train_test_split(
    X_train_hybrid_full,
    y_train_full,
    test_size=0.2,
    random_state=SEED,
    stratify=y_train_full
)

history = final_mlp.fit(
    X_train_final, y_train_final,
    validation_data=(X_val_final, y_val_final),
    epochs=80,
    batch_size=128,
    callbacks=[early_stop],
    verbose=1
)

mlp_val_probs = final_mlp.predict(X_val_final, verbose=0).ravel()
best_mlp_threshold = 0.50
best_mlp_f1 = -1

for th in candidate_thresholds:
    preds = (mlp_val_probs >= th).astype(int)
    score = f1_score(y_val_final, preds, zero_division=0)
    if score > best_mlp_f1:
        best_mlp_f1 = score
        best_mlp_threshold = th

print(f"\nBest MLP threshold: {best_mlp_threshold:.2f}")

mlp_test_probs = final_mlp.predict(X_test_hybrid, verbose=0).ravel()
mlp_test_pred = (mlp_test_probs >= best_mlp_threshold).astype(int)

mlp_test_metrics = compute_metrics(y_test, mlp_test_pred, mlp_test_probs)
print_metrics("FINAL TEST: HYBRID AIS + PSO + MLP", mlp_test_metrics)
plot_confusion_matrix(mlp_test_metrics['cm'], "Confusion Matrix - Final MLP")
add_results_row("MLP", mlp_test_metrics)

# ---------- Final Random Forest ----------
final_rf = clone(rf_model)
final_rf.fit(X_train_hybrid_full, y_train_full)

rf_val_model = clone(rf_model)
rf_val_model.fit(X_train_final, y_train_final)
rf_val_probs = rf_val_model.predict_proba(X_val_final)[:, 1]

best_rf_threshold = 0.50
best_rf_f1 = -1

for th in candidate_thresholds:
    preds = (rf_val_probs >= th).astype(int)
    score = f1_score(y_val_final, preds, zero_division=0)
    if score > best_rf_f1:
        best_rf_f1 = score
        best_rf_threshold = th

print(f"Best RF threshold: {best_rf_threshold:.2f}")

rf_test_probs = final_rf.predict_proba(X_test_hybrid)[:, 1]
rf_test_pred = (rf_test_probs >= best_rf_threshold).astype(int)

rf_test_metrics = compute_metrics(y_test, rf_test_pred, rf_test_probs)
print_metrics("FINAL TEST: HYBRID AIS + PSO + RANDOM FOREST", rf_test_metrics)
plot_confusion_matrix(rf_test_metrics['cm'], "Confusion Matrix - Final Random Forest")
add_results_row("Random Forest", rf_test_metrics)

# ---------- Final XGBoost ----------
final_xgb = clone(xgb_model)
final_xgb.fit(X_train_hybrid_full, y_train_full)

xgb_val_model = clone(xgb_model)
xgb_val_model.fit(X_train_final, y_train_final)
xgb_val_probs = xgb_val_model.predict_proba(X_val_final)[:, 1]

best_xgb_threshold = 0.50
best_xgb_f1 = -1

for th in candidate_thresholds:
    preds = (xgb_val_probs >= th).astype(int)
    score = f1_score(y_val_final, preds, zero_division=0)
    if score > best_xgb_f1:
        best_xgb_f1 = score
        best_xgb_threshold = th

print(f"Best XGB threshold: {best_xgb_threshold:.2f}")

xgb_test_probs = final_xgb.predict_proba(X_test_hybrid)[:, 1]
xgb_test_pred = (xgb_test_probs >= best_xgb_threshold).astype(int)

xgb_test_metrics = compute_metrics(y_test, xgb_test_pred, xgb_test_probs)
print_metrics("FINAL TEST: HYBRID AIS + PSO + XGBOOST", xgb_test_metrics)
plot_confusion_matrix(xgb_test_metrics['cm'], "Confusion Matrix - Final XGBoost")
add_results_row("XGBoost", xgb_test_metrics)

# =========================================================
# Step 24: Final Ensemble on Test Set
# =========================================================
# Hard voting
ensemble_votes = mlp_test_pred + rf_test_pred + xgb_test_pred
hard_ensemble_test_pred = (ensemble_votes >= 2).astype(int)
hard_ensemble_test_probs = (mlp_test_probs + rf_test_probs + xgb_test_probs) / 3.0

hard_test_metrics = compute_metrics(y_test, hard_ensemble_test_pred, hard_ensemble_test_probs)
print_metrics("FINAL TEST: HARD VOTING ENSEMBLE", hard_test_metrics)
plot_confusion_matrix(hard_test_metrics['cm'], "Confusion Matrix - Final Hard Voting Ensemble")
add_results_row("Hard Voting Ensemble", hard_test_metrics)

# Soft voting
soft_test_probs = (
    0.15 * mlp_test_probs +
    0.30 * rf_test_probs +
    0.55 * xgb_test_probs
)

soft_val_probs = (
    0.15 * mlp_val_probs +
    0.30 * rf_val_probs +
    0.55 * xgb_val_probs
)

best_soft_threshold = 0.50
best_soft_f1 = -1

for th in candidate_thresholds:
    preds = (soft_val_probs >= th).astype(int)
    score = f1_score(y_val_final, preds, zero_division=0)
    if score > best_soft_f1:
        best_soft_f1 = score
        best_soft_threshold = th

print(f"\nBest Soft Voting threshold: {best_soft_threshold:.2f}")

soft_ensemble_test_pred = (soft_test_probs >= best_soft_threshold).astype(int)

soft_test_metrics = compute_metrics(y_test, soft_ensemble_test_pred, soft_test_probs)
print_metrics("FINAL TEST: SOFT VOTING ENSEMBLE", soft_test_metrics)
plot_confusion_matrix(soft_test_metrics['cm'], "Confusion Matrix - Final Soft Voting Ensemble")
add_results_row("Soft Voting Ensemble", soft_test_metrics)

# =========================================================
# Step 25: ROC Curve (Final Test)
# =========================================================
plt.figure(figsize=(8, 6))
plt.plot(mlp_test_metrics['roc_fpr'], mlp_test_metrics['roc_tpr'], label=f"MLP (AUC={mlp_test_metrics['roc_auc']:.4f})", linewidth=2)
plt.plot(rf_test_metrics['roc_fpr'], rf_test_metrics['roc_tpr'], label=f"RF (AUC={rf_test_metrics['roc_auc']:.4f})", linewidth=2)
plt.plot(xgb_test_metrics['roc_fpr'], xgb_test_metrics['roc_tpr'], label=f"XGB (AUC={xgb_test_metrics['roc_auc']:.4f})", linewidth=2)
plt.plot(hard_test_metrics['roc_fpr'], hard_test_metrics['roc_tpr'], label=f"Hard Ensemble (AUC={hard_test_metrics['roc_auc']:.4f})", linewidth=2)
plt.plot(soft_test_metrics['roc_fpr'], soft_test_metrics['roc_tpr'], label=f"Soft Ensemble (AUC={soft_test_metrics['roc_auc']:.4f})", linewidth=2)
plt.plot([0, 1], [0, 1], linestyle='--', linewidth=1)
plt.title("ROC Curve - Final Test Set", fontsize=13, fontweight='bold')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.tight_layout()
plt.show()

# =========================================================
# Step 26: Precision-Recall Curve (Final Test)
# =========================================================
plt.figure(figsize=(8, 6))
plt.plot(mlp_test_metrics['pr_recall'], mlp_test_metrics['pr_precision'], label=f"MLP (AUC={mlp_test_metrics['pr_auc']:.4f})", linewidth=2)
plt.plot(rf_test_metrics['pr_recall'], rf_test_metrics['pr_precision'], label=f"RF (AUC={rf_test_metrics['pr_auc']:.4f})", linewidth=2)
plt.plot(xgb_test_metrics['pr_recall'], xgb_test_metrics['pr_precision'], label=f"XGB (AUC={xgb_test_metrics['pr_auc']:.4f})", linewidth=2)
plt.plot(hard_test_metrics['pr_recall'], hard_test_metrics['pr_precision'], label=f"Hard Ensemble (AUC={hard_test_metrics['pr_auc']:.4f})", linewidth=2)
plt.plot(soft_test_metrics['pr_recall'], soft_test_metrics['pr_precision'], label=f"Soft Ensemble (AUC={soft_test_metrics['pr_auc']:.4f})", linewidth=2)
plt.title("Precision-Recall Curve - Final Test Set", fontsize=13, fontweight='bold')
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.legend()
plt.tight_layout()
plt.show()

# =========================================================
# Step 27: Feature Importance from PSO
# =========================================================
selected_mask = np.zeros(num_features)
selected_mask[selected_feature_indices] = 1

pso_importance_df = pd.DataFrame({
    'Feature': X_train_base.columns,
    'Selected_by_PSO': selected_mask.astype(int)
}).sort_values(by='Selected_by_PSO', ascending=False)

top_selected_features = pd.DataFrame({
    'Feature': selected_feature_names,
    'Selection_Order': np.arange(1, len(selected_feature_names) + 1)
})

plt.figure(figsize=(10, max(5, len(selected_feature_names) * 0.25)))
sns.barplot(
    data=top_selected_features,
    y='Feature',
    x='Selection_Order',
    orient='h'
)
plt.title("Feature Importance from PSO (Selected Features)", fontsize=13, fontweight='bold')
plt.xlabel("Selection Rank")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()

# =========================================================
# Step 28: PSO Fitness Convergence Curve
# =========================================================
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(gbest_history) + 1), gbest_history, linewidth=2)
plt.title("PSO Fitness Convergence", fontsize=13, fontweight='bold')
plt.xlabel("Iteration")
plt.ylabel("Best Fitness")
plt.tight_layout()
plt.show()

# =========================================================
# Step 29: CV Summary Tables
# =========================================================
print("\nMLP 10-Fold CV Results")
print(mlp_cv_df.round(4))

print("\nRF 10-Fold CV Results")
print(rf_cv_df.round(4))

print("\nXGB 10-Fold CV Results")
print(xgb_cv_df.round(4))

cv_summary_df = pd.DataFrame([
    {
        'Model': 'MLP',
        'Accuracy_mean': mlp_cv_mean['Accuracy'],
        'Precision_mean': mlp_cv_mean['Precision'],
        'Recall_mean': mlp_cv_mean['Recall'],
        'F1_mean': mlp_cv_mean['F1-score'],
        'Balanced_Accuracy_mean': mlp_cv_mean['Balanced Accuracy'],
        'ROC_AUC_mean': mlp_cv_mean['ROC-AUC'],
        'PR_AUC_mean': mlp_cv_mean['PR-AUC']
    },
    {
        'Model': 'Random Forest',
        'Accuracy_mean': rf_cv_mean['Accuracy'],
        'Precision_mean': rf_cv_mean['Precision'],
        'Recall_mean': rf_cv_mean['Recall'],
        'F1_mean': rf_cv_mean['F1-score'],
        'Balanced_Accuracy_mean': rf_cv_mean['Balanced Accuracy'],
        'ROC_AUC_mean': rf_cv_mean['ROC-AUC'],
        'PR_AUC_mean': rf_cv_mean['PR-AUC']
    },
    {
        'Model': 'XGBoost',
        'Accuracy_mean': xgb_cv_mean['Accuracy'],
        'Precision_mean': xgb_cv_mean['Precision'],
        'Recall_mean': xgb_cv_mean['Recall'],
        'F1_mean': xgb_cv_mean['F1-score'],
        'Balanced_Accuracy_mean': xgb_cv_mean['Balanced Accuracy'],
        'ROC_AUC_mean': xgb_cv_mean['ROC-AUC'],
        'PR_AUC_mean': xgb_cv_mean['PR-AUC']
    },
    {
        'Model': 'Hard Voting Ensemble (OOF)',
        'Accuracy_mean': hard_cv_metrics['accuracy'],
        'Precision_mean': hard_cv_metrics['precision'],
        'Recall_mean': hard_cv_metrics['recall'],
        'F1_mean': hard_cv_metrics['f1'],
        'Balanced_Accuracy_mean': hard_cv_metrics['balanced_accuracy'],
        'ROC_AUC_mean': hard_cv_metrics['roc_auc'],
        'PR_AUC_mean': hard_cv_metrics['pr_auc']
    },
    {
        'Model': 'Soft Voting Ensemble (OOF)',
        'Accuracy_mean': soft_cv_metrics['accuracy'],
        'Precision_mean': soft_cv_metrics['precision'],
        'Recall_mean': soft_cv_metrics['recall'],
        'F1_mean': soft_cv_metrics['f1'],
        'Balanced_Accuracy_mean': soft_cv_metrics['balanced_accuracy'],
        'ROC_AUC_mean': soft_cv_metrics['roc_auc'],
        'PR_AUC_mean': soft_cv_metrics['pr_auc']
    }
])

print("\n10-FOLD CV SUMMARY")
print(cv_summary_df.round(4))

# =========================================================
# Step 30: Final Test Results Summary Table
# =========================================================
results_df = pd.DataFrame(results_summary)
print("\nFINAL TEST RESULTS SUMMARY")
print(results_df.round(4))

# =========================================================
# Step 31: Save Results
# =========================================================
output_dir = '/content/hybrid_ids_outputs_cv'
os.makedirs(output_dir, exist_ok=True)

selected_features_df = pd.DataFrame({
    "Selected_Features": selected_feature_names
})
selected_features_df.to_csv(f'{output_dir}/hybrid_selected_features.csv', index=False)

ais_feature_names = [
    'min_dist', 'mean_dist', 'std_dist', 'median_dist',
    'closest3_mean', 'closest5_mean', 'closest10_mean',
    'activated_08', 'activated_10', 'activated_15',
    'max_affinity', 'mean_affinity', 'std_affinity'
]

train_ais_df = pd.DataFrame(X_train_ais_full, columns=ais_feature_names)
train_ais_df.to_csv(f'{output_dir}/train_ais_scores_full.csv', index=False)

pred_df = pd.DataFrame({
    'Actual': y_test,
    'MLP_Prob': mlp_test_probs,
    'MLP_Pred': mlp_test_pred,
    'RF_Prob': rf_test_probs,
    'RF_Pred': rf_test_pred,
    'XGB_Prob': xgb_test_probs,
    'XGB_Pred': xgb_test_pred,
    'Hard_Ensemble_Pred': hard_ensemble_test_pred,
    'Soft_Prob': soft_test_probs,
    'Soft_Ensemble_Pred': soft_ensemble_test_pred
})
pred_df.to_csv(f'{output_dir}/hybrid_ensemble_predictions.csv', index=False)

results_df.to_csv(f'{output_dir}/final_test_model_results_summary.csv', index=False)
cv_summary_df.to_csv(f'{output_dir}/cv_summary_results.csv', index=False)
mlp_cv_df.to_csv(f'{output_dir}/mlp_10fold_results.csv', index=False)
rf_cv_df.to_csv(f'{output_dir}/rf_10fold_results.csv', index=False)
xgb_cv_df.to_csv(f'{output_dir}/xgb_10fold_results.csv', index=False)
pso_importance_df.to_csv(f'{output_dir}/pso_feature_importance.csv', index=False)

print("\nSaved files:")
print(f"{output_dir}/hybrid_selected_features.csv")
print(f"{output_dir}/train_ais_scores_full.csv")
print(f"{output_dir}/hybrid_ensemble_predictions.csv")
print(f"{output_dir}/final_test_model_results_summary.csv")
print(f"{output_dir}/cv_summary_results.csv")
print(f"{output_dir}/mlp_10fold_results.csv")
print(f"{output_dir}/rf_10fold_results.csv")
print(f"{output_dir}/xgb_10fold_results.csv")
print(f"{output_dir}/pso_feature_importance.csv")
```
