```python
# =========================================================
# HYBRID IDS: PSO + NEGATIVE SELECTION AIS + RF + ISO FOREST + MLP + XGBOOST + ENSEMBLE
# Dataset: CICIoT2023
# Google Colab Full Implementation
# MULTICLASS + SMOTETomek + Two-Tier + ADR + FPR + TNR + SHAP
# =========================================================

# =========================================================
# Step 0: Install Required Libraries
# =========================================================
!pip install -q imbalanced-learn xgboost shap

# =========================================================
# Step 1: Import Libraries
# =========================================================
import os
import random
import warnings
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore")

from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, IsolationForest
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    balanced_accuracy_score
)

from imblearn.combine import SMOTETomek
from imblearn.over_sampling import SMOTE

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping

from xgboost import XGBClassifier
import shap

SEED = 42
os.environ["PYTHONHASHSEED"] = str(SEED)
np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)

sns.set(style="whitegrid")

print("✅ Libraries imported successfully!")

# Optional: if Colab GPU causes CUDA error, uncomment this BEFORE TensorFlow init
# os.environ["CUDA_VISIBLE_DEVICES"] = "-1"

# =========================================================
# Step 2: Mount Google Drive
# =========================================================
from google.colab import drive
drive.mount('/content/drive')

# =========================================================
# Step 3: Load CICIoT2023 Dataset
# =========================================================
dataset_path = '/content/drive/MyDrive/Computational_Intelligence/CICIoT2023'
csv_files = sorted([f for f in os.listdir(dataset_path) if f.endswith('.csv')])

if len(csv_files) == 0:
    raise FileNotFoundError("No CSV files found in the CICIoT2023 folder.")

print("Available CSV files:", csv_files[:10])

# Load first CSV to keep structure close to your original code
df = pd.read_csv(os.path.join(dataset_path, csv_files[0]), low_memory=False)

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist()[:20])

if 'label' not in df.columns:
    raise ValueError("Dataset must contain a 'label' column.")

# =========================================================
# Step 3.1: Remove Ultra-Rare Classes for Stable SMOTE
# =========================================================
min_samples_required = 6

label_counts = df['label'].value_counts()
valid_labels = label_counts[label_counts >= min_samples_required].index
removed_labels = label_counts[label_counts < min_samples_required].index.tolist()

print("\nRemoving ultra-rare classes:", removed_labels)

df = df[df['label'].isin(valid_labels)].reset_index(drop=True)

print("Dataset shape after removing ultra-rare classes:", df.shape)
print("Remaining class distribution:")
print(df['label'].value_counts())

# =========================================================
# Step 3.2: Train/Test Split
# =========================================================
train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    random_state=SEED,
    stratify=df['label']
)

train_df = train_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

print("\nTrain shape:", train_df.shape)
print("Test shape :", test_df.shape)

# =========================================================
# Step 4: Clean Labels and Optional Columns
# =========================================================
if 'difficulty' in train_df.columns:
    train_df = train_df.drop(columns=['difficulty'])
if 'difficulty' in test_df.columns:
    test_df = test_df.drop(columns=['difficulty'])

if 'label' not in train_df.columns:
    train_df.rename(columns={train_df.columns[-1]: 'label'}, inplace=True)
    test_df.rename(columns={test_df.columns[-1]: 'label'}, inplace=True)

train_df['label'] = train_df['label'].astype(str).str.strip()
test_df['label'] = test_df['label'].astype(str).str.strip()

# Detect normal / benign labels
train_unique_lower = set(train_df['label'].str.lower().unique().tolist())
test_unique_lower = set(test_df['label'].str.lower().unique().tolist())
all_unique_lower = train_unique_lower.union(test_unique_lower)

normal_keywords = ['normal', 'normal.', 'benign', 'benigntraffic']
normal_labels = [
    lbl for lbl in all_unique_lower
    if lbl in normal_keywords or 'benign' in lbl or 'normal' in lbl
]

if len(normal_labels) == 0:
    fallback_normal = train_df['label'].str.lower().value_counts().idxmax()
    normal_labels = [fallback_normal]
    print(f"⚠️ No explicit benign/normal label found. Using most frequent label as normal: {fallback_normal}")

print("Detected normal labels:", normal_labels)

# Binary helper only for AIS / Isolation Forest
train_df['binary_helper'] = train_df['label'].str.lower().apply(lambda x: 0 if x in normal_labels else 1)
test_df['binary_helper'] = test_df['label'].str.lower().apply(lambda x: 0 if x in normal_labels else 1)

print("\nTrain helper binary distribution:")
print(train_df['binary_helper'].value_counts())

print("\nTest helper binary distribution:")
print(test_df['binary_helper'].value_counts())

# Multiclass label encoding
label_encoder = LabelEncoder()
all_labels = pd.concat([train_df['label'], test_df['label']], axis=0).astype(str)
label_encoder.fit(all_labels)

train_df['multi_label'] = label_encoder.transform(train_df['label'])
test_df['multi_label'] = label_encoder.transform(test_df['label'])

class_names = list(label_encoder.classes_)
num_classes = len(class_names)
all_class_ids = np.arange(num_classes)

print("\nNumber of classes:", num_classes)
print("Classes:", class_names)

normal_encoded = [label_encoder.transform([lbl])[0] for lbl in label_encoder.classes_ if lbl.lower() in normal_labels]
normal_encoded_set = set(normal_encoded)

print("Encoded normal classes:", normal_encoded)

# =========================================================
# Step 5: Encode Categorical Features
# =========================================================
combined_df = pd.concat(
    [
        train_df.drop(columns=['binary_helper', 'multi_label']),
        test_df.drop(columns=['binary_helper', 'multi_label'])
    ],
    axis=0
).reset_index(drop=True)

categorical_cols = combined_df.select_dtypes(include=['object']).columns.tolist()
categorical_cols = [col for col in categorical_cols if col != 'label']

print("\nCategorical columns:", categorical_cols)

for col in categorical_cols:
    le = LabelEncoder()
    combined_df[col] = le.fit_transform(combined_df[col].astype(str))

train_processed = combined_df.iloc[:len(train_df)].copy()
test_processed = combined_df.iloc[len(train_df):].copy()

# =========================================================
# Step 6: Separate Features and Labels
# =========================================================
X_train_full = train_processed.drop(columns=['label'])
X_test_full = test_processed.drop(columns=['label'])

y_train_full = train_df['multi_label'].values
y_test = test_df['multi_label'].values

y_train_binary_helper = train_df['binary_helper'].values
y_test_binary_helper = test_df['binary_helper'].values

X_train_full = X_train_full.select_dtypes(include=[np.number]).copy()
X_test_full = X_test_full.select_dtypes(include=[np.number]).copy()

# =========================================================
# Step 6.1: Remove Infinite and Missing Values
# =========================================================
X_train_full = X_train_full.replace([np.inf, -np.inf], np.nan)
X_test_full = X_test_full.replace([np.inf, -np.inf], np.nan)

train_mask = ~X_train_full.isna().any(axis=1)
test_mask = ~X_test_full.isna().any(axis=1)

X_train_full = X_train_full[train_mask].reset_index(drop=True)
y_train_full = y_train_full[train_mask]
y_train_binary_helper = y_train_binary_helper[train_mask]

X_test_full = X_test_full[test_mask].reset_index(drop=True)
y_test = y_test[test_mask]
y_test_binary_helper = y_test_binary_helper[test_mask]

# Remove constant columns
constant_cols = [col for col in X_train_full.columns if X_train_full[col].nunique() <= 1]
if len(constant_cols) > 0:
    X_train_full = X_train_full.drop(columns=constant_cols)
    X_test_full = X_test_full.drop(columns=constant_cols, errors='ignore')
    print("\nRemoved constant columns:", constant_cols)

# =========================================================
# Step 6.2: Improve Features
# =========================================================
def add_engineered_features(df_in):
    df_out = df_in.copy()
    df_out['feat_mean'] = df_out.mean(axis=1)
    df_out['feat_std'] = df_out.std(axis=1)
    df_out['feat_max'] = df_out.max(axis=1)
    df_out['feat_min'] = df_out.min(axis=1)
    df_out['feat_range'] = df_out['feat_max'] - df_out['feat_min']
    return df_out

X_train_full = add_engineered_features(X_train_full)
X_test_full = add_engineered_features(X_test_full)

print("\nFeature shape after engineering:")
print("Train:", X_train_full.shape)
print("Test :", X_test_full.shape)

# =========================================================
# Step 7: Standardization
# =========================================================
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_full)
X_test_scaled = scaler.transform(X_test_full)

X_train_full = pd.DataFrame(X_train_scaled, columns=X_train_full.columns)
X_test_full = pd.DataFrame(X_test_scaled, columns=X_test_full.columns)

# =========================================================
# Step 8: Train / Validation Split
# =========================================================
(
    X_train, X_val,
    y_train, y_val,
    y_train_binary, y_val_binary
) = train_test_split(
    X_train_full,
    y_train_full,
    y_train_binary_helper,
    test_size=0.2,
    random_state=SEED,
    stratify=y_train_full
)

X_train = X_train.reset_index(drop=True)
X_val = X_val.reset_index(drop=True)
X_test_full = X_test_full.reset_index(drop=True)

print("\nTrain:", X_train.shape)
print("Val  :", X_val.shape)
print("Test :", X_test_full.shape)

# =========================================================
# Step 8.5: Apply SMOTE + Tomek Links (Fixed)
# =========================================================
print("\nApplying SMOTE + Tomek Links on training set...")

class_counts = Counter(y_train)
print("Class counts before SMOTETomek:")
print(class_counts)

min_class_size = min(class_counts.values())

if min_class_size < 2:
    raise ValueError(
        f"At least one class has only {min_class_size} sample. "
        f"SMOTE cannot work. Remove ultra-rare classes first."
    )

safe_k_neighbors = min(5, min_class_size - 1)
print("Using SMOTE k_neighbors =", safe_k_neighbors)

smote = SMOTE(
    k_neighbors=safe_k_neighbors,
    random_state=SEED
)

smote_tomek = SMOTETomek(
    smote=smote,
    random_state=SEED
)

X_train_res, y_train_res = smote_tomek.fit_resample(X_train, y_train)

print("Before SMOTETomek:", X_train.shape, len(y_train))
print("After  SMOTETomek:", X_train_res.shape, len(y_train_res))

# rebuild binary helper after resampling
y_train_binary_res = np.array([0 if cls in normal_encoded_set else 1 for cls in y_train_res])

X_train = pd.DataFrame(X_train_res, columns=X_train.columns)
y_train = np.asarray(y_train_res)
y_train_binary = np.asarray(y_train_binary_res)

# =========================================================
# Step 9: Balanced Subset for PSO
# =========================================================
def create_balanced_subset_multiclass(X, y, samples_per_class=4000):
    chosen_indices = []

    for cls in np.unique(y):
        cls_idx = np.where(y == cls)[0]
        n_cls = min(len(cls_idx), samples_per_class)
        chosen_cls = np.random.choice(cls_idx, n_cls, replace=False)
        chosen_indices.extend(chosen_cls.tolist())

    chosen_indices = np.array(chosen_indices)
    np.random.shuffle(chosen_indices)

    X_sub = X.iloc[chosen_indices].reset_index(drop=True)
    y_sub = y[chosen_indices]
    return X_sub, y_sub

X_pso, y_pso = create_balanced_subset_multiclass(X_train, y_train, samples_per_class=4000)

# =========================================================
# Step 10: PSO Feature Selection using RF Fitness
# =========================================================
X_pso_np = X_pso.values
X_val_np = X_val.values

def fitness_function(particle, X_train_sub, X_val_full, y_train_sub, y_val, alpha=0.97):
    selected = np.where(particle == 1)[0]

    if len(selected) == 0:
        return 0.0

    Xtr = X_train_sub[:, selected]
    Xv = X_val_full[:, selected]

    clf = RandomForestClassifier(
        n_estimators=120,
        random_state=SEED,
        n_jobs=-1
    )
    clf.fit(Xtr, y_train_sub)
    preds = clf.predict(Xv)

    bal_acc = balanced_accuracy_score(y_val, preds)
    feature_ratio = len(selected) / X_train_sub.shape[1]
    fitness = alpha * bal_acc + (1 - alpha) * (1 - feature_ratio)
    return fitness

num_features = X_pso_np.shape[1]
num_particles = 20
num_iterations = 25
w, c1, c2 = 0.72, 1.49, 1.49

particles = np.random.randint(0, 2, size=(num_particles, num_features))
velocities = np.random.uniform(-1, 1, size=(num_particles, num_features))

for i in range(num_particles):
    if particles[i].sum() == 0:
        particles[i, np.random.randint(0, num_features)] = 1

pbest_positions = particles.copy()
pbest_scores = np.zeros(num_particles)
gbest_position = None
gbest_score = -1
gbest_history = []

for i in range(num_particles):
    score = fitness_function(particles[i], X_pso_np, X_val_np, y_pso, y_val)
    pbest_scores[i] = score
    if score > gbest_score:
        gbest_score = score
        gbest_position = particles[i].copy()

print("\nStarting PSO...")
for iteration in range(num_iterations):
    for i in range(num_particles):
        r1 = np.random.rand(num_features)
        r2 = np.random.rand(num_features)

        velocities[i] = (
            w * velocities[i]
            + c1 * r1 * (pbest_positions[i] - particles[i])
            + c2 * r2 * (gbest_position - particles[i])
        )

        sigmoid = 1 / (1 + np.exp(-velocities[i]))
        particles[i] = (np.random.rand(num_features) < sigmoid).astype(int)

        if particles[i].sum() == 0:
            particles[i, np.random.randint(0, num_features)] = 1

        score = fitness_function(particles[i], X_pso_np, X_val_np, y_pso, y_val)

        if score > pbest_scores[i]:
            pbest_scores[i] = score
            pbest_positions[i] = particles[i].copy()

        if score > gbest_score:
            gbest_score = score
            gbest_position = particles[i].copy()

    gbest_history.append(gbest_score)
    print(f"Iteration {iteration + 1}/{num_iterations}, Best Fitness: {gbest_score:.4f}")

selected_feature_indices = np.where(gbest_position == 1)[0]
selected_feature_names = X_train.columns[selected_feature_indices]

print("\nSelected feature count:", len(selected_feature_indices))
print("Selected features:", list(selected_feature_names))

X_train_sel = X_train.values[:, selected_feature_indices]
X_val_sel = X_val.values[:, selected_feature_indices]
X_test_sel = X_test_full.values[:, selected_feature_indices]

# =========================================================
# Step 11: Balanced AIS Training Data
# =========================================================
def create_balanced_numpy_subset(X, y_binary, normal_count=5000, attack_count=5000):
    normal_idx = np.where(y_binary == 0)[0]
    attack_idx = np.where(y_binary == 1)[0]

    n_normal = min(len(normal_idx), normal_count)
    n_attack = min(len(attack_idx), attack_count)

    chosen_normal = np.random.choice(normal_idx, n_normal, replace=False)
    chosen_attack = np.random.choice(attack_idx, n_attack, replace=False)

    return X[chosen_normal], X[chosen_attack]

self_samples, attack_samples = create_balanced_numpy_subset(
    X_train_sel, y_train_binary, normal_count=5000, attack_count=5000
)

print("\nSelf samples :", self_samples.shape)
print("Attack samples:", attack_samples.shape)

# =========================================================
# Step 12: Negative Selection Detector Generation
# =========================================================
def generate_attack_detectors(self_samples, attack_samples, num_detectors=600, min_self_distance=0.6):
    detectors = []
    attempts = 0
    max_attempts = num_detectors * 100

    while len(detectors) < num_detectors and attempts < max_attempts:
        base = attack_samples[np.random.randint(0, len(attack_samples))].copy()
        candidate = base + np.random.normal(0, 0.08, size=base.shape)

        d_self = np.min(np.linalg.norm(self_samples - candidate, axis=1))

        if d_self >= min_self_distance:
            detectors.append(candidate)

        attempts += 1

    return np.array(detectors)

detectors = generate_attack_detectors(
    self_samples=self_samples,
    attack_samples=attack_samples,
    num_detectors=1000,
    min_self_distance=0.6
)

print("\nDetector shape:", detectors.shape)

# =========================================================
# Step 13: Create AIS Anomaly Features
# =========================================================
def create_ais_scores(X, detectors):
    feature_list = []

    for sample in X:
        dists = np.linalg.norm(detectors - sample, axis=1)
        sorted_dists = np.sort(dists)

        min_dist = np.min(dists)
        mean_dist = np.mean(dists)
        std_dist = np.std(dists)
        median_dist = np.median(dists)

        closest3_mean = np.mean(sorted_dists[:3]) if len(sorted_dists) >= 3 else np.mean(sorted_dists)
        closest5_mean = np.mean(sorted_dists[:5]) if len(sorted_dists) >= 5 else np.mean(sorted_dists)
        closest10_mean = np.mean(sorted_dists[:10]) if len(sorted_dists) >= 10 else np.mean(sorted_dists)

        activated_08 = np.sum(dists <= 0.8)
        activated_10 = np.sum(dists <= 1.0)
        activated_15 = np.sum(dists <= 1.5)

        affinities = 1 / (1 + dists)
        max_affinity = np.max(affinities)
        mean_affinity = np.mean(affinities)
        std_affinity = np.std(affinities)

        feature_list.append([
            min_dist,
            mean_dist,
            std_dist,
            median_dist,
            closest3_mean,
            closest5_mean,
            closest10_mean,
            activated_08,
            activated_10,
            activated_15,
            max_affinity,
            mean_affinity,
            std_affinity
        ])

    return np.array(feature_list)

X_train_ais = create_ais_scores(X_train_sel, detectors)
X_val_ais = create_ais_scores(X_val_sel, detectors)
X_test_ais = create_ais_scores(X_test_sel, detectors)

print("\nAIS feature shape (train):", X_train_ais.shape)

# =========================================================
# Step 14: Merge Selected Features + AIS Features
# =========================================================
X_train_hybrid = np.hstack((X_train_sel, X_train_ais))
X_val_hybrid = np.hstack((X_val_sel, X_val_ais))
X_test_hybrid = np.hstack((X_test_sel, X_test_ais))

print("Hybrid train shape:", X_train_hybrid.shape)
print("Hybrid val shape  :", X_val_hybrid.shape)
print("Hybrid test shape :", X_test_hybrid.shape)

X_train_hybrid = np.asarray(X_train_hybrid).astype(np.float32)
X_val_hybrid = np.asarray(X_val_hybrid).astype(np.float32)
X_test_hybrid = np.asarray(X_test_hybrid).astype(np.float32)

y_train = np.asarray(y_train).astype(np.int32)
y_val = np.asarray(y_val).astype(np.int32)
y_test = np.asarray(y_test).astype(np.int32)

hybrid_feature_names = list(selected_feature_names) + [
    'min_dist', 'mean_dist', 'std_dist', 'median_dist',
    'closest3_mean', 'closest5_mean', 'closest10_mean',
    'activated_08', 'activated_10', 'activated_15',
    'max_affinity', 'mean_affinity', 'std_affinity'
]

# =========================================================
# Step 15: Evaluation + Visualization Helpers
# =========================================================
results_summary = []

def compute_attack_metrics(y_true, y_pred, normal_encoded_labels):
    actual_normal_mask = np.isin(y_true, normal_encoded_labels)
    actual_attack_mask = ~actual_normal_mask

    total_normal = np.sum(actual_normal_mask)
    total_attack = np.sum(actual_attack_mask)

    pred_normal_on_normal = np.sum(np.isin(y_pred[actual_normal_mask], normal_encoded_labels))
    pred_attack_on_normal = total_normal - pred_normal_on_normal

    pred_attack_on_attack = np.sum(~np.isin(y_pred[actual_attack_mask], normal_encoded_labels))

    adr = pred_attack_on_attack / total_attack if total_attack > 0 else 0.0
    fpr = pred_attack_on_normal / total_normal if total_normal > 0 else 0.0
    tnr = pred_normal_on_normal / total_normal if total_normal > 0 else 0.0

    return adr, fpr, tnr

def compute_metrics_multiclass(y_true, y_pred, normal_encoded_labels):
    cm = confusion_matrix(y_true, y_pred, labels=all_class_ids)

    accuracy = accuracy_score(y_true, y_pred)
    precision_macro = precision_score(y_true, y_pred, average='macro', zero_division=0)
    recall_macro = recall_score(y_true, y_pred, average='macro', zero_division=0)
    f1_macro = f1_score(y_true, y_pred, average='macro', zero_division=0)

    precision_weighted = precision_score(y_true, y_pred, average='weighted', zero_division=0)
    recall_weighted = recall_score(y_true, y_pred, average='weighted', zero_division=0)
    f1_weighted = f1_score(y_true, y_pred, average='weighted', zero_division=0)

    bal_acc = balanced_accuracy_score(y_true, y_pred)
    adr, fpr, tnr = compute_attack_metrics(y_true, y_pred, normal_encoded_labels)

    return {
        'cm': cm,
        'accuracy': accuracy,
        'precision_macro': precision_macro,
        'recall_macro': recall_macro,
        'f1_macro': f1_macro,
        'precision_weighted': precision_weighted,
        'recall_weighted': recall_weighted,
        'f1_weighted': f1_weighted,
        'balanced_accuracy': bal_acc,
        'ADR': adr,
        'FPR': fpr,
        'TNR': tnr
    }

def print_metrics(name, metrics_dict):
    print(f"\n{'='*80}")
    print(name)
    print(f"{'='*80}")
    print(f"Accuracy             : {metrics_dict['accuracy']:.4f}")
    print(f"Precision (Macro)    : {metrics_dict['precision_macro']:.4f}")
    print(f"Recall (Macro)       : {metrics_dict['recall_macro']:.4f}")
    print(f"F1-score (Macro)     : {metrics_dict['f1_macro']:.4f}")
    print(f"Precision (Weighted) : {metrics_dict['precision_weighted']:.4f}")
    print(f"Recall (Weighted)    : {metrics_dict['recall_weighted']:.4f}")
    print(f"F1-score (Weighted)  : {metrics_dict['f1_weighted']:.4f}")
    print(f"Balanced Accuracy    : {metrics_dict['balanced_accuracy']:.4f}")
    print(f"ADR                  : {metrics_dict['ADR']:.4f}")
    print(f"FPR                  : {metrics_dict['FPR']:.4f}")
    print(f"TNR                  : {metrics_dict['TNR']:.4f}")
    print("\nConfusion Matrix:")
    print(metrics_dict['cm'])

def plot_confusion_matrix(cm, title, class_names):
    plt.figure(figsize=(12, 10))
    sns.heatmap(
        cm,
        annot=False,
        fmt='d',
        cmap='Blues',
        xticklabels=class_names,
        yticklabels=class_names
    )
    plt.title(title, fontsize=13, fontweight='bold')
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")
    plt.xticks(rotation=90)
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()

def add_results_row(name, metrics_dict):
    results_summary.append({
        'Model': name,
        'Accuracy': metrics_dict['accuracy'],
        'Precision_Macro': metrics_dict['precision_macro'],
        'Recall_Macro': metrics_dict['recall_macro'],
        'F1_Macro': metrics_dict['f1_macro'],
        'Precision_Weighted': metrics_dict['precision_weighted'],
        'Recall_Weighted': metrics_dict['recall_weighted'],
        'F1_Weighted': metrics_dict['f1_weighted'],
        'Balanced_Accuracy': metrics_dict['balanced_accuracy'],
        'ADR': metrics_dict['ADR'],
        'FPR': metrics_dict['FPR'],
        'TNR': metrics_dict['TNR']
    })

# =========================================================
# Step 16: Train MLP
# =========================================================
mlp_model = Sequential([
    Input(shape=(X_train_hybrid.shape[1],)),
    Dense(256, activation='relu'),
    Dropout(0.4),
    Dense(128, activation='relu'),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(num_classes, activation='softmax')
])

mlp_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("\nMLP MODEL SUMMARY")
mlp_model.summary()

early_stop = EarlyStopping(
    monitor='val_loss',
    patience=6,
    restore_best_weights=True
)

history = mlp_model.fit(
    X_train_hybrid, y_train,
    validation_data=(X_val_hybrid, y_val),
    epochs=80,
    batch_size=128,
    callbacks=[early_stop],
    verbose=1
)

# =========================================================
# Step 17: Training vs Validation Loss
# =========================================================
plt.figure(figsize=(8, 5))
plt.plot(history.history['loss'], label='Training Loss', linewidth=2)
plt.plot(history.history['val_loss'], label='Validation Loss', linewidth=2)
plt.title('Training vs Validation Loss (MLP)', fontsize=13, fontweight='bold')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.tight_layout()
plt.show()

# =========================================================
# Step 18: MLP Prediction
# =========================================================
mlp_test_probs = mlp_model.predict(X_test_hybrid, verbose=0)
mlp_pred = np.argmax(mlp_test_probs, axis=1)

mlp_metrics = compute_metrics_multiclass(y_test, mlp_pred, normal_encoded)
print_metrics("HYBRID AIS + PSO + MLP", mlp_metrics)
plot_confusion_matrix(mlp_metrics['cm'], "Confusion Matrix - MLP", class_names)
add_results_row("MLP", mlp_metrics)

# =========================================================
# Step 19: Train Random Forest (Tier 1)
# =========================================================
rf_model = RandomForestClassifier(
    n_estimators=400,
    max_depth=None,
    random_state=SEED,
    n_jobs=-1
)

rf_model.fit(X_train_hybrid, y_train)

rf_test_probs = rf_model.predict_proba(X_test_hybrid)
rf_pred = np.argmax(rf_test_probs, axis=1)

rf_metrics = compute_metrics_multiclass(y_test, rf_pred, normal_encoded)
print_metrics("HYBRID AIS + PSO + RANDOM FOREST", rf_metrics)
plot_confusion_matrix(rf_metrics['cm'], "Confusion Matrix - Random Forest", class_names)
add_results_row("Random Forest", rf_metrics)

# =========================================================
# Step 19.1: SHAP Analysis for Random Forest
# =========================================================
print("\nRunning SHAP analysis for Random Forest...")

rf_shap_df = None
try:
    rf_sample_size = min(300, len(X_test_hybrid))
    rf_sample_idx = np.random.choice(len(X_test_hybrid), rf_sample_size, replace=False)
    X_rf_shap = X_test_hybrid[rf_sample_idx]

    rf_explainer = shap.TreeExplainer(rf_model)
    rf_shap_values = rf_explainer.shap_values(X_rf_shap)

    if isinstance(rf_shap_values, list):
        rf_abs_mean = np.mean([np.abs(sv) for sv in rf_shap_values], axis=0)
        rf_global_importance = np.mean(rf_abs_mean, axis=0)
    else:
        rf_abs_mean = np.abs(rf_shap_values)
        if len(rf_abs_mean.shape) == 3:
            rf_global_importance = np.mean(rf_abs_mean, axis=(0, 2))
        else:
            rf_global_importance = np.mean(rf_abs_mean, axis=0)

    rf_shap_df = pd.DataFrame({
        'Feature': hybrid_feature_names,
        'MeanAbsSHAP': rf_global_importance
    }).sort_values(by='MeanAbsSHAP', ascending=False)

    print("\nTop 20 RF SHAP Features:")
    print(rf_shap_df.head(20))

    plt.figure(figsize=(10, 8))
    sns.barplot(
        data=rf_shap_df.head(20),
        x='MeanAbsSHAP',
        y='Feature',
        orient='h'
    )
    plt.title("Top 20 SHAP Features - Random Forest", fontsize=13, fontweight='bold')
    plt.xlabel("Mean |SHAP value|")
    plt.ylabel("Feature")
    plt.tight_layout()
    plt.show()

except Exception as e:
    print("RF SHAP could not be completed:", str(e))

# =========================================================
# Step 20: Train XGBoost
# =========================================================
xgb_model = XGBClassifier(
    n_estimators=900,
    max_depth=10,
    learning_rate=0.03,
    subsample=0.90,
    colsample_bytree=0.90,
    gamma=0.1,
    min_child_weight=2,
    objective='multi:softprob',
    num_class=num_classes,
    eval_metric='mlogloss',
    random_state=SEED,
    n_jobs=-1
)

xgb_model.fit(
    X_train_hybrid, y_train,
    eval_set=[(X_val_hybrid, y_val)],
    verbose=False
)

xgb_test_probs = xgb_model.predict_proba(X_test_hybrid)
xgb_pred = np.argmax(xgb_test_probs, axis=1)

xgb_metrics = compute_metrics_multiclass(y_test, xgb_pred, normal_encoded)
print_metrics("HYBRID AIS + PSO + XGBOOST", xgb_metrics)
plot_confusion_matrix(xgb_metrics['cm'], "Confusion Matrix - XGBoost", class_names)
add_results_row("XGBoost", xgb_metrics)

# =========================================================
# Step 20.1: SHAP Analysis for XGBoost
# =========================================================
print("\nRunning SHAP analysis for XGBoost...")

xgb_shap_df = None
try:
    xgb_sample_size = min(300, len(X_test_hybrid))
    xgb_sample_idx = np.random.choice(len(X_test_hybrid), xgb_sample_size, replace=False)
    X_xgb_shap = X_test_hybrid[xgb_sample_idx]

    xgb_explainer = shap.TreeExplainer(xgb_model)
    xgb_shap_values = xgb_explainer.shap_values(X_xgb_shap)

    if isinstance(xgb_shap_values, list):
        xgb_abs_mean = np.mean([np.abs(sv) for sv in xgb_shap_values], axis=0)
        xgb_global_importance = np.mean(xgb_abs_mean, axis=0)
    else:
        xgb_abs_mean = np.abs(xgb_shap_values)
        if len(xgb_abs_mean.shape) == 3:
            xgb_global_importance = np.mean(xgb_abs_mean, axis=(0, 2))
        else:
            xgb_global_importance = np.mean(xgb_abs_mean, axis=0)

    xgb_shap_df = pd.DataFrame({
        'Feature': hybrid_feature_names,
        'MeanAbsSHAP': xgb_global_importance
    }).sort_values(by='MeanAbsSHAP', ascending=False)

    print("\nTop 20 XGBoost SHAP Features:")
    print(xgb_shap_df.head(20))

    plt.figure(figsize=(10, 8))
    sns.barplot(
        data=xgb_shap_df.head(20),
        x='MeanAbsSHAP',
        y='Feature',
        orient='h'
    )
    plt.title("Top 20 SHAP Features - XGBoost", fontsize=13, fontweight='bold')
    plt.xlabel("Mean |SHAP value|")
    plt.ylabel("Feature")
    plt.tight_layout()
    plt.show()

except Exception as e:
    print("XGBoost SHAP could not be completed:", str(e))

# =========================================================
# Step 21: Ensemble Voting (Hard)
# =========================================================
ensemble_votes = np.vstack([mlp_pred, rf_pred, xgb_pred]).T

def majority_vote(row):
    values, counts = np.unique(row, return_counts=True)
    return values[np.argmax(counts)]

ensemble_pred = np.apply_along_axis(majority_vote, axis=1, arr=ensemble_votes)

ensemble_metrics = compute_metrics_multiclass(y_test, ensemble_pred, normal_encoded)
print_metrics("HYBRID AIS + PSO + ENSEMBLE VOTING", ensemble_metrics)
plot_confusion_matrix(ensemble_metrics['cm'], "Confusion Matrix - Hard Voting Ensemble", class_names)
add_results_row("Hard Voting Ensemble", ensemble_metrics)

# =========================================================
# Step 22: Soft Voting
# =========================================================
soft_probs = (
    0.15 * mlp_test_probs +
    0.30 * rf_test_probs +
    0.55 * xgb_test_probs
)

soft_ensemble_pred = np.argmax(soft_probs, axis=1)

soft_metrics = compute_metrics_multiclass(y_test, soft_ensemble_pred, normal_encoded)
print_metrics("HYBRID AIS + PSO + SOFT VOTING ENSEMBLE", soft_metrics)
plot_confusion_matrix(soft_metrics['cm'], "Confusion Matrix - Soft Voting Ensemble", class_names)
add_results_row("Soft Voting Ensemble", soft_metrics)

# =========================================================
# Step 23: Two-Tier System
# Tier 1: Random Forest
# Tier 2: Isolation Forest on RF-benign predictions
# =========================================================
print("\nTraining Isolation Forest (Tier 2)...")

normal_idx = np.where(y_train_binary == 0)[0]
X_normal_for_iso = X_train_sel[normal_idx]

iso_model = IsolationForest(
    n_estimators=200,
    contamination=0.01,
    random_state=SEED,
    n_jobs=-1
)

iso_model.fit(X_normal_for_iso)

rf_pred_labels_lower = np.array([class_names[idx].lower() for idx in rf_pred])
rf_pred_is_benign = np.array([1 if lbl in normal_labels else 0 for lbl in rf_pred_labels_lower])

benign_pred_indices = np.where(rf_pred_is_benign == 1)[0]

iso_flags = np.ones(len(y_test), dtype=int)

if len(benign_pred_indices) > 0:
    iso_subset = X_test_sel[benign_pred_indices]
    iso_scores = iso_model.predict(iso_subset)  # 1 normal, -1 anomaly
    iso_flags[benign_pred_indices] = iso_scores

two_tier_pred = rf_pred.copy()

# if RF says benign but IF says anomaly,
# redirect to best non-benign RF class
for idx in benign_pred_indices:
    if iso_flags[idx] == -1:
        probs = rf_test_probs[idx].copy()
        for bcls in normal_encoded_set:
            probs[bcls] = -1
        two_tier_pred[idx] = np.argmax(probs)

two_tier_metrics = compute_metrics_multiclass(y_test, two_tier_pred, normal_encoded)
print_metrics("TWO-TIER SYSTEM (RF + Isolation Forest)", two_tier_metrics)
plot_confusion_matrix(two_tier_metrics['cm'], "Confusion Matrix - Two-Tier System", class_names)
add_results_row("Two-Tier RF + IF", two_tier_metrics)

# =========================================================
# Step 24: Results Summary Table
# =========================================================
results_df = pd.DataFrame(results_summary)
print("\nFINAL RESULTS SUMMARY")
print(results_df.round(4))

# =========================================================
# Step 25: Feature Importance from PSO
# =========================================================
selected_mask = np.zeros(num_features)
selected_mask[selected_feature_indices] = 1

pso_importance_df = pd.DataFrame({
    'Feature': X_train.columns,
    'Selected_by_PSO': selected_mask.astype(int)
}).sort_values(by='Selected_by_PSO', ascending=False)

top_selected_features = pd.DataFrame({
    'Feature': selected_feature_names,
    'Selection_Order': np.arange(1, len(selected_feature_names) + 1)
})

plt.figure(figsize=(10, max(5, len(selected_feature_names) * 0.25)))
sns.barplot(
    data=top_selected_features,
    y='Feature',
    x='Selection_Order',
    orient='h'
)
plt.title("Feature Importance from PSO (Selected Features)", fontsize=13, fontweight='bold')
plt.xlabel("Selection Rank")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()

# =========================================================
# Step 26: PSO Fitness Convergence Curve
# =========================================================
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(gbest_history) + 1), gbest_history, linewidth=2)
plt.title("PSO Fitness Convergence", fontsize=13, fontweight='bold')
plt.xlabel("Iteration")
plt.ylabel("Best Fitness")
plt.tight_layout()
plt.show()

# =========================================================
# Step 27: Classification Reports
# =========================================================
print("\nClassification Report - MLP")
print(classification_report(y_test, mlp_pred, target_names=class_names, zero_division=0))

print("\nClassification Report - RF")
print(classification_report(y_test, rf_pred, target_names=class_names, zero_division=0))

print("\nClassification Report - XGBoost")
print(classification_report(y_test, xgb_pred, target_names=class_names, zero_division=0))

print("\nClassification Report - Hard Voting Ensemble")
print(classification_report(y_test, ensemble_pred, target_names=class_names, zero_division=0))

print("\nClassification Report - Soft Voting Ensemble")
print(classification_report(y_test, soft_ensemble_pred, target_names=class_names, zero_division=0))

print("\nClassification Report - Two-Tier RF + Isolation Forest")
print(classification_report(y_test, two_tier_pred, target_names=class_names, zero_division=0))

# =========================================================
# Step 28: Save Results
# =========================================================
output_dir = '/content/hybrid_ids_outputs_multiclass_two_tier'
os.makedirs(output_dir, exist_ok=True)

selected_features_df = pd.DataFrame({
    "Selected_Features": selected_feature_names
})
selected_features_df.to_csv(f'{output_dir}/hybrid_selected_features.csv', index=False)

ais_feature_names = [
    'min_dist', 'mean_dist', 'std_dist', 'median_dist',
    'closest3_mean', 'closest5_mean', 'closest10_mean',
    'activated_08', 'activated_10', 'activated_15',
    'max_affinity', 'mean_affinity', 'std_affinity'
]

train_ais_df = pd.DataFrame(X_train_ais, columns=ais_feature_names)
train_ais_df.to_csv(f'{output_dir}/train_ais_scores.csv', index=False)

pred_df = pd.DataFrame({
    'Actual': y_test,
    'MLP_Pred': mlp_pred,
    'RF_Pred': rf_pred,
    'XGB_Pred': xgb_pred,
    'Hard_Ensemble_Pred': ensemble_pred,
    'Soft_Ensemble_Pred': soft_ensemble_pred,
    'Two_Tier_Pred': two_tier_pred
})

for i, cls_name in enumerate(class_names):
    safe_name = str(cls_name).replace("/", "_").replace(" ", "_")
    pred_df[f"MLP_Prob_{safe_name}"] = mlp_test_probs[:, i]
    pred_df[f"RF_Prob_{safe_name}"] = rf_test_probs[:, i]
    pred_df[f"XGB_Prob_{safe_name}"] = xgb_test_probs[:, i]
    pred_df[f"Soft_Prob_{safe_name}"] = soft_probs[:, i]

pred_df.to_csv(f'{output_dir}/hybrid_ensemble_predictions_multiclass.csv', index=False)

results_df.to_csv(f'{output_dir}/model_results_summary_multiclass.csv', index=False)
pso_importance_df.to_csv(f'{output_dir}/pso_feature_importance.csv', index=False)

label_map_df = pd.DataFrame({
    'Encoded_Label': np.arange(num_classes),
    'Original_Label': class_names
})
label_map_df.to_csv(f'{output_dir}/label_mapping.csv', index=False)

if rf_shap_df is not None:
    rf_shap_df.to_csv(f'{output_dir}/rf_shap_feature_importance.csv', index=False)

if xgb_shap_df is not None:
    xgb_shap_df.to_csv(f'{output_dir}/xgb_shap_feature_importance.csv', index=False)

print("\nSaved files:")
print(f"{output_dir}/hybrid_selected_features.csv")
print(f"{output_dir}/train_ais_scores.csv")
print(f"{output_dir}/hybrid_ensemble_predictions_multiclass.csv")
print(f"{output_dir}/model_results_summary_multiclass.csv")
print(f"{output_dir}/pso_feature_importance.csv")
print(f"{output_dir}/label_mapping.csv")
if rf_shap_df is not None:
    print(f"{output_dir}/rf_shap_feature_importance.csv")
if xgb_shap_df is not None:
    print(f"{output_dir}/xgb_shap_feature_importance.csv")
```
